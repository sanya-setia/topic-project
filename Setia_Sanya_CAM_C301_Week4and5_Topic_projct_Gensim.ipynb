```python
from joblib import load
from gensim.test.utils import common_texts
from gensim.corpora.dictionary import Dictionary
from gensim.models.ldamodel import LdaModel
```

```python
combined_docs = load('combined_docs.joblib')
```

```python
# Tokenise the documents.
tokenised_docs = [doc.lower().split() for doc in combined_docs]
```

```python
# Create a dictionary from the documents
dictionary = Dictionary(tokenised_docs)

# Filter out extremes 
# Remove words that appear in less than 2 documents or more than 50% of documents
dictionary.filter_extremes(no_below=2, no_above=0.5)

# Recreate corpus with filtered dictionary
corpus = [dictionary.doc2bow(doc) for doc in tokenised_docs]
```

## Build the LDA Model using Dictionary and Corpus

```python
num_topics = 10
passes = 20


lda_model = LdaModel(corpus=corpus, id2word=dictionary, num_topics=num_topics, passes=passes)
```

```python
for idx, topic in lda_model.print_topics(-1):
    print("Topic: {} \nWords: {}".format(idx, topic))
    print("\n")
```

```python
import pyLDAvis.gensim_models as gensimvis
import pyLDAvis

# Prepare the visualisation.
pyLDAvis.enable_notebook()
vis = gensimvis.prepare(lda_model, corpus, dictionary)
pyLDAvis.display(vis)
```
